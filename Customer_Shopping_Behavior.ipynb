```python
import pandas as pd
import numpy as np
```

**CSV FILE**

```python
df = pd.read_csv("/content/customer_shopping_behavior.csv", sep="\t")
df.head()
```

**Here CSV File Information, DataType, Details.**

```python
df.info()
df.describe()
df.describe(include="all")
df.shape
df.columns
```

**Checking The Null Value.**

```python
df.isnull().sum()
```

**Set The Null Value.**

```python
df["Review Rating"] = df.groupby("Category")["Review Rating"].transform(lambda x:x.fillna(x.median()))
```

**Updatet The Nul Values.**

```python
df.isnull().sum()
```

**Replace The column Name (Space to UnderScore).**

```python
# df.columns = df.columns.str.lower()
df.columns = df.columns.str.replace(' -','_')
df.columns
```

**Replace The column Name(Perchase Amount USD to Perchase Amount).**

```python
df.columns = df.columns.str.lower().str.replace(r"[- ]+", "_", regex=True)
df = df.rename(columns={'purchase_amount_(usd)':'purchase_amount'})
df.columns
```

**Age Band According to the Age.**

```python
# Age Group
labels = ['Young','Adult', 'Middle_age','Senior']
df['age_group'] = pd.qcut(df['age'], q = 4, labels=labels)
df[['age', 'age_group']].head(5)
```

**Purchase Frequency Days**

```python
# Purchase Frequency Days

frequency_mapping = {
    'Fortnightly': 14,
    'Weekly': 7,
    'Monthly': 30,
    'Quarterly': 90,
    'Bi-Weekly': 14,
    'Annually': 365,
    'Every 3 Months': 90
}
df['purchase_frequency_days'] = df['frequency_of_purchases'].map(frequency_mapping)

df[['purchase_frequency_days', 'frequency_of_purchases']].head(5)

```
